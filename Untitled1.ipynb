```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from keras.models import Sequential
from keras.layers import Dense, Dropout, BatchNormalization
from keras import optimizers
from keras import initializers

# 1. Автономно генеруємо набір даних для класифікації
X_raw, y_raw = make_classification(
    n_samples=2000,      # Кількість рядків
    n_features=14,       # Кількість ознак
    n_informative=10,    # Кількість інформативних ознак
    n_classes=2,         # Бінарна класифікація (наприклад, дохід >50К чи ні)
    random_state=42
)

# Перетворюємо у формат Pandas DataFrame
X = pd.DataFrame(X_raw, columns=[f'feature_{i}' for i in range(14)])
Y = pd.get_dummies(y_raw)  # Dummy-кодування для виходу мережі (як у методичці)

# 2. Розбиваємо на навчальну (85%) та тестову (15%) вибірки
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.15, random_state=42)

# 3. Масштабування (стандартизація) даних для стабільності оптимізатора
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)
X_train = (X_train - mean) / std
X_test = (X_test - mean) / std

print("Генерація та підготовка даних пройшла успішно!")
print(f"Форма навчальних даних: {X_train.shape}")
print(f"Форма тестових даних: {X_test.shape}")
```

```python
# Базова конфігурація
model_base = Sequential()
model_base.add(Dense(10, input_dim=X_train.shape[1], activation="relu"))
model_base.add(Dense(5, activation="relu"))
model_base.add(Dense(Y.shape[1], activation="softmax"))

model_base.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("--- Навчання базової моделі ---")
history_base = model_base.fit(X_train, Y_train, epochs=40, validation_split=0.15, batch_size=16, verbose=1)
```

```python
# Налаштування зворотного виклику EarlyStopping (зупинка навчання при стабілізації помилки)
early_stop = tf.keras.callbacks.EarlyStopping(monitor='val_accuracy', patience=6, restore_best_weights=True)

# Параметри ініціалізації ваг для Варіанту 10 (TruncatedNormal з фіксованим seed)
init_w = initializers.TruncatedNormal(mean=0., stddev=0.05, seed=101010)
init_b = initializers.Constant(1e-3)

# Створення оптимізованої архітектури
model_opt = Sequential()

# Перший прихований шар + L2 регуляризація ваг + Пакетна нормалізація + Dropout
model_opt.add(Dense(12, input_dim=X_train.shape[1], activation="relu",
                    kernel_initializer=init_w, bias_initializer=init_b, kernel_regularizer='l2'))
model_opt.add(BatchNormalization())
model_opt.add(Dropout(0.25))

# Другий прихований шар + L2 регуляризація ваг + Пакетна нормалізація + Dropout
model_opt.add(Dense(8, activation="relu",
                    kernel_initializer=init_w, bias_initializer=init_b, kernel_regularizer='l2'))
model_opt.add(BatchNormalization())
model_opt.add(Dropout(0.2))

# Вихідний шар
model_opt.add(Dense(Y.shape[1], activation="softmax", kernel_initializer=init_w, bias_initializer=init_b))

# Налаштування оптимізатора Adam зі зміненою швидкістю навчання (Learning Rate)
custom_adam = optimizers.Adam(learning_rate=0.004)
model_opt.compile(optimizer=custom_adam, loss='categorical_crossentropy', metrics=['accuracy'])

print("\n--- Навчання оптимізованої моделі Юрія (Варіант 10) ---")
history_opt = model_opt.fit(X_train, Y_train, epochs=50, validation_split=0.15, batch_size=16,
                            callbacks=[early_stop], verbose=1)

# Порівняльна оцінка результатів на тестових даних
_, acc_b = model_base.evaluate(X_test, Y_test, verbose=0)
_, acc_o = model_opt.evaluate(X_test, Y_test, verbose=0)
print("\n" + "="*50)
print(f"[РЕЗУЛЬТАТ ЮРІЯ] Точність базової моделі: {acc_b*100:.2f}%")
print(f"[РЕЗУЛЬТАТ ЮРІЯ] Точність оптимізованої моделі: {acc_o*100:.2f}%")
print("="*50)
```

```python
plt.figure(figsize=(14, 5))

# Графік для базової моделі
plt.subplot(1, 2, 1)
plt.plot(history_base.history['accuracy'], label='Навчання (Train)')
plt.plot(history_base.history['val_accuracy'], label='Валідація (Val)')
plt.title('Базова модель: Динаміка точності')
plt.xlabel('Епоха')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# Графік для твоєї оптимізованої моделі (Варіант 10)
plt.subplot(1, 2, 2)
plt.plot(history_opt.history['accuracy'], label='Навчання (Train)', color='green')
plt.plot(history_opt.history['val_accuracy'], label='Валідація (Val)', color='orange')
plt.title('Оптимізована модель (Варіант 10): Динаміка точності')
plt.xlabel('Епоха')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

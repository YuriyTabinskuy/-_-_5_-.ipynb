```python
import math

def find_sum_of_squares(n):

    if not isinstance(n, int):
        raise TypeError("Вхідні дані мають бути тільки цілим числом (int)!")
    if n < 0:
        raise ValueError("Число повинно бути невід'ємним (більше або дорівнює 0)!")

    pairs = []

    limit = int(math.isqrt(n))

    for a in range(limit + 1):
        b_squared = n - a**2
        b = math.isqrt(b_squared)
        if b**2 == b_squared:

            if a <= b:
                pairs.append((a, b))

    return pairs
```

```python


print("=== Тест 1: Число 50 (очікуємо варіанти) ===")
num_1 = 50
res_1 = find_sum_of_squares(num_1)
print(f"Число {num_1} як сума квадратів: {res_1}")


print("\n=== Тест 2: Число 25 (очікуємо варіанти) ===")
num_2 = 25
res_2 = find_sum_of_squares(num_2)
print(f"Число {num_2} як сума квадратів: {res_2}")


print("\n=== Тест 3: Число 7 (неможливо розкласти) ===")
num_3 = 7
res_3 = find_sum_of_squares(num_3)
print(f"Число {num_3} як сума квадратів: {res_3}")

print("\n=== Тест 7 (Контроль помилок): Введення некоректного типу ===")
try:
    find_sum_of_squares("п'ятдесят")
except Exception as e:
    print(f"Зловлено помилку типу (валідація працює): {e}")

try:
    find_sum_of_squares(-10)
except Exception as e:
    print(f"Зловлено помилку значення (валідація працює): {e}")
```

```python
import matplotlib.pyplot as plt

numbers_to_test = list(range(0, 101))
ways_count = []

for num in numbers_to_test:
    combinations = find_sum_of_squares(num)
    ways_count.append(len(combinations))


plt.figure(figsize=(12, 5))
plt.stem(numbers_to_test, ways_count, linefmt='b-', markerfmt='bo', basefmt='r-')

plt.title("Кількість способів подання чисел від 0 до 100 як суми двох квадратів", fontsize=14, fontweight='bold')
plt.xlabel("Число (n)", fontsize=11)
plt.ylabel("Кількість комбінацій (a² + b² = n)", fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)


for idx, count in enumerate(ways_count):
    if count >= 2:
        plt.annotate(f"n={idx}", (idx, count), textcoords="offset points", xytext=(0,10), ha='center', fontweight='bold', color='green')


output_image_path = '/content/drive/MyDrive/Colab_Labs/Lab1/sum_of_squares_chart.png'
plt.savefig(output_image_path, dpi=300)
print(f"Графік успішно збережено за шляхом: {output_image_path}")

plt.show()
```

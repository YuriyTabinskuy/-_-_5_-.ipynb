```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from google.colab import files

print("Потрібно завантажити файл лабораторної.")
# Цей рядок відкриє вікно для вибору файлу на твоєму комп'ютері
uploaded = files.upload()

# Отримуємо назву завантаженого файлу
file_name = list(uploaded.keys())[0]
print(f"✅ Файл {file_name} успішно завантажено в Colab!")

# Читаємо завантажений файл
df = pd.read_csv(file_name)

print("\n--- Перші 5 записів таблиці (Завдання 2) ---")
display(df.head())

print("\n--- Баланс класів у стовпці 'gender' (Завдання 5) ---")
# Якщо у твоєму файлі стовпець називається трохи інакше (наприклад, з великої літери),
# код автоматично підлаштується під потрібну назву стовпця зі статтю
gender_col = [col for col in df.columns if 'gender' in col.lower()][0]
print(df[gender_col].value_counts(normalize=True))
```

```python
# Виділяємо предиктори та цільову змінну
X = df.drop('gender', axis=1)

# Автоматична факторизація, якщо раптом є текст у характеристиках
text_cols = X.select_dtypes(include=['object']).columns.tolist()
if text_cols:
    X[text_cols] = X[text_cols].apply(lambda col: pd.factorize(col, sort=True)[0])

# Перетворення міток класів у формат One Hot Encoding
Y = pd.get_dummies(df['gender'], prefix='gender')

# Розбиття на навчальну та тестову вибірки (85% на 15%)
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.15, random_state=42)

# Стандартизація даних
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)
X_train = (X_train - mean) / std
X_test = (X_test - mean) / std

# Побудова архітектури нейромережі
model = Sequential([
    Dense(12, input_dim=X_train.shape[1], activation="relu"),
    Dense(6, activation="relu"),
    Dense(2, activation="softmax") # 2 класи: Male / Female
])

# Компіляція
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
print("Модель успішно створено та зкомпільовано!")
```

```python
# Навчання (записуємо історію)
history = model.fit(X_train, Y_train,
                    epochs=40,
                    validation_split=0.1,
                    batch_size=16,
                    verbose=1)

print("\n=== ОЦІНКА НА ТЕСТОВІЙ ВИБІРЦІ ===")
loss, accuracy = model.evaluate(X_test, Y_test, verbose=0)
print(f"Тестові втрати (Loss): {loss:.4f}")
print(f"Тестова точність (Accuracy): {accuracy*100:.2f}%\n")

# Малюємо графік точності
plt.figure(figsize=(8, 5))
plt.plot(history.history['accuracy'], label='Навчальна точність (Train)')
if 'val_accuracy' in history.history:
    plt.plot(history.history['val_accuracy'], label='Валідаційна точність (Val)')
plt.title('Залежність точності від епохи навчання')
plt.xlabel('Епоха')
plt.ylabel('Точність (Accuracy)')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()

print("\n=== ПЕРЕВІРКА НА ОДНОМУ ЗРАЗКУ ===")
# Беремо перший зразок з тестів
sample = X_test[0:1]
prediction = model.predict(sample, verbose=0)

pred_class_idx = np.argmax(prediction)
true_class_idx = np.argmax(Y_test[0:1].values)

class_labels = Y.columns.tolist()
predicted_label = class_labels[pred_class_idx].replace('gender_', '')
true_label = class_labels[true_class_idx].replace('gender_', '')
probability = np.max(prediction) * 100

print(f"Передбачено клас: {predicted_label} з достовірністю {probability:.2f}%")
print(f"Повинно бути (Правильна відповідь): {true_label}")
```

```python

!pip install deap

import random
import numpy as np
from deap import base, creator, tools, algorithms


WEIGHTS = [10, 20, 33, 14, 7, 5, 12, 19, 25, 15]
VALUES = [100, 120, 150, 90, 60, 40, 75, 80, 110, 65]
MAX_KNAPSACK_WEIGHT = 40

print("=== Постановка задачі ===")
print(f"Ваги предметів: {WEIGHTS}")
print(f"Цінності предметів: {VALUES}")
print(f"Максимальна дозволена вага рюкзака: {MAX_KNAPSACK_WEIGHT} кг\n")


def knapsack_eval(individual):
    total_weight = sum(ind * w for ind, w in zip(individual, WEIGHTS))
    total_value = sum(ind * v for ind, v in zip(individual, VALUES))

    if total_weight > MAX_KNAPSACK_WEIGHT:
        return 0,
    return total_value,


if "FitnessMax" not in creator.__dict__:
    creator.create("FitnessMax", base.Fitness, weights=(1.0,))
if "IndividualKnapsack" not in creator.__dict__:
    creator.create("IndividualKnapsack", list, fitness=creator.FitnessMax)

tb = base.Toolbox()
tb.register("attr_bin", random.randint, 0, 1)
tb.register("individual", tools.initRepeat, creator.IndividualKnapsack, tb.attr_bin, n=len(WEIGHTS))
tb.register("population", tools.initRepeat, list, tb.individual)

tb.register("evaluate", knapsack_eval)
tb.register("mate", tools.cxTwoPoint)
tb.register("mutate", tools.mutFlipBit, indpb=0.1)
tb.register("select", tools.selTournament, tournsize=3)

population = tb.population(n=100)
hof = tools.HallOfFame(1)

stats = tools.Statistics(lambda ind: ind.fitness.values)
stats.register("max", np.max)

print("=== Еволюційний процес ===")
population, log = algorithms.eaSimple(population, tb, cxpb=0.8, mutpb=0.2, ngen=30, stats=stats, halloffame=hof, verbose=True)

best_sol = hof[0]
best_weight = sum(ind * w for ind, w in zip(best_sol, WEIGHTS))
best_value = sum(ind * v for ind, v in zip(best_sol, VALUES))

print("\n=== Результат вирішення задачі ===")
print(f"Найкраща комбінація (0/1): {best_sol}")
print(f"Загальна отримана цінність: {best_value}")
print(f"Загальна вага рюкзака: {best_weight} кг (Максимум: {MAX_KNAPSACK_WEIGHT} кг)")
```
